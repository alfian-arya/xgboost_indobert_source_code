## Cell 3 — Visualisasi Perbandingan

```python
fig, axes = plt.subplots(1,3,figsize=(16,5))
fig.suptitle('Perbandingan 12 Eksperimen — Klasifikasi Komentar Media Sosial',
             fontweight='bold', fontsize=12)

colors_model={'LR':'#888780','XGB-NoTune':'#7F77DD',
              'XGB-Tuned':'#534AB7','IndoBERT':'#1D9E75'}

df_A_man = df_res[df_res['ske']=='A-Manual']
df_A_gab = df_res[df_res['ske']=='A-Gabungan']
df_B_gab = df_res[df_res['ske']=='B-Gabungan']

for ax,(df_s,ske_title) in zip(axes,[
    (df_A_man,'A-Manual (Test=3.707)'),
    (df_A_gab,'A-Gabungan (Test=14.027)'),
    (df_B_gab,'B-Gabungan (Test=3.707 fix)')]):
    bars=ax.bar(df_s['model'],df_s['f1'],
               color=[colors_model[m] for m in df_s['model']],
               edgecolor='white',width=0.6)
    for bar,v in zip(bars,df_s['f1']):
        ax.text(bar.get_x()+bar.get_width()/2,bar.get_height()+0.005,
                f'{v:.4f}',ha='center',va='bottom',fontsize=9,fontweight='bold')
    ax.set_title(ske_title,fontweight='bold')
    ax.set_ylabel('Macro F1'); ax.set_ylim(0,1.1)
    ax.tick_params(axis='x',rotation=20)
    ax.grid(axis='y',alpha=0.3)
    ax.spines['top'].set_visible(False); ax.spines['right'].set_visible(False)

from matplotlib.patches import Patch
legend=[Patch(facecolor=v,label=k) for k,v in colors_model.items()]
fig.legend(handles=legend,loc='lower center',ncol=4,bbox_to_anchor=(0.5,-0.05))
plt.tight_layout()
plt.savefig(os.path.join(RESULT_DIR,'perbandingan_12_eksperimen.png'),bbox_inches='tight')
plt.close(); print('Plot tersimpan!')
```
