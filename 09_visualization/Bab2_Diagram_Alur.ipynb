## Setup & Konfigurasi

```python
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch
import matplotlib.patheffects as pe
import numpy as np
import os

OUT = 'output_bab2'
os.makedirs(OUT, exist_ok=True)

# ── Global style (Draw.io-like, Arial) ──────────────────────────────────────
plt.rcParams.update({
    'font.family'      : 'Arial',
    'font.size'        : 10,
    'figure.dpi'       : 200,
    'savefig.dpi'      : 200,
    'savefig.bbox'     : 'tight',
    'savefig.facecolor': 'white',
    'savefig.pad_inches': 0.15,
})

# ── Warna Draw.io ────────────────────────────────────────────────────────────
C = {
    'box_fill'  : '#FFFFFF',
    'box_edge'  : '#23445D',
    'box_shadow': '#D0D8E0',
    'start_fill': '#1E4D78',
    'start_txt' : '#FFFFFF',
    'proc_fill' : '#DAE8FC',
    'proc_edge' : '#6C8EBF',
    'io_fill'   : '#D5E8D4',
    'io_edge'   : '#82B366',
    'decs_fill' : '#FFF2CC',
    'decs_edge' : '#D6B656',
    'bert_fill' : '#E1D5E7',
    'bert_edge' : '#9673A6',
    'arrow'     : '#23445D',
    'dash'      : '#7F7F7F',
    'group_bg'  : '#F5F9FF',
    'group_edge': '#B0C4DE',
    'txt'       : '#1A1A1A',
    'txt_light' : '#555555',
}

# ── Helper: kotak Draw.io ────────────────────────────────────────────────────
def box(ax, cx, cy, w, h, txt, style='proc', fontsize=9, bold=False, latex=None):
    fc = C[f'{style}_fill']; ec = C[f'{style}_edge']
    lw = 1.2
    rect = FancyBboxPatch((cx-w/2, cy-h/2), w, h,
                          boxstyle='round,pad=0.015',
                          facecolor=fc, edgecolor=ec, linewidth=lw, zorder=3)
    ax.add_patch(rect)
    if latex:
        ax.text(cx, cy+0.045, txt, ha='center', va='center',
                fontsize=fontsize, fontweight='bold' if bold else 'normal',
                color=C['txt'], fontfamily='Arial', zorder=4)
        ax.text(cx, cy-0.055, latex, ha='center', va='center',
                fontsize=fontsize-1, color='#1D5FA3', fontfamily='DejaVu Sans',
                style='italic', zorder=4)
    else:
        fw = 'bold' if bold else 'normal'
        ax.text(cx, cy, txt, ha='center', va='center',
                fontsize=fontsize, fontweight=fw,
                color=C['start_txt'] if style=='start' else C['txt'],
                fontfamily='Arial', multialignment='center', zorder=4)

def start_box(ax, cx, cy, w, h, txt, fontsize=9):
    rect = FancyBboxPatch((cx-w/2, cy-h/2), w, h,
                          boxstyle='round,pad=0.04',
                          facecolor=C['start_fill'], edgecolor=C['start_fill'],
                          linewidth=0, zorder=3)
    ax.add_patch(rect)
    ax.text(cx, cy, txt, ha='center', va='center',
            fontsize=fontsize, fontweight='bold',
            color=C['start_txt'], fontfamily='Arial', zorder=4)

def grp(ax, x1, y1, x2, y2, label=''):
    from matplotlib.patches import Rectangle
    rect = Rectangle((x1, y1), x2-x1, y2-y1,
                     facecolor=C['group_bg'], edgecolor=C['group_edge'],
                     linewidth=0.8, linestyle='--', zorder=1)
    ax.add_patch(rect)
    if label:
        ax.text(x1+0.02, y2-0.03, label, ha='left', va='top',
                fontsize=7.5, color=C['group_edge'], fontstyle='italic',
                fontfamily='Arial', zorder=2)

def arr(ax, x1, y1, x2, y2, dashed=False, label=''):
    ls = (0, (4,3)) if dashed else '-'
    color = C['dash'] if dashed else C['arrow']
    ax.annotate('', xy=(x2,y2), xytext=(x1,y1),
                arrowprops=dict(arrowstyle='->', color=color, lw=1.0,
                                linestyle=ls,
                                connectionstyle='arc3,rad=0'),
                zorder=5)
    if label:
        mx,my = (x1+x2)/2,(y1+y2)/2
        ax.text(mx+0.02, my, label, ha='left', va='center',
                fontsize=7.5, color=C['dash'], fontfamily='Arial')

def save(fig, name):
    path = f'{OUT}/{name}'
    fig.savefig(path)
    plt.close(fig)
    print(f'✅ Tersimpan: {path}')

print('Setup selesai. Siap membuat diagram.')
```
