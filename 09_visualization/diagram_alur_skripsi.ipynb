```python
# ============================================================
# DIAGRAM ALUR FORMAT SKRIPSI — Gambar 1–9 Bab 2
# Seprianto — Teknik Informatika Universitas Hasanuddin
# Jalankan: python diagram_alur_skripsi.py
# Output  : folder output_gambar/ berisi PNG siap insert Word
# ============================================================

import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch
import numpy as np
import os

OUT = "output_gambar"
os.makedirs(OUT, exist_ok=True)

plt.rcParams.update({
    'font.family'      : 'Arial',
    'figure.dpi'       : 180,
    'savefig.dpi'      : 180,
    'savefig.bbox'     : 'tight',
    'savefig.facecolor': 'white',
})

# ── Warna ─────────────────────────────────────────────────────────────────────
C_BOX   = '#FFFFFF'
C_EDGE  = '#000000'
C_ARROW = '#000000'
C_SHADE = '#E8E8F0'
C_DASH  = '#555555'
C_TXT   = '#000000'

# ── Helper: kotak teks ────────────────────────────────────────────────────────
def draw_box(ax, x, y, w, h, text, fontsize=9, bold=False,
             fc=C_BOX, ec=C_EDGE, lw=0.8, style='square'):
    box = FancyBboxPatch((x - w/2, y - h/2), w, h,
                         boxstyle='round,pad=0.02' if style=='round' else 'square,pad=0',
                         facecolor=fc, edgecolor=ec, linewidth=lw)
    ax.add_patch(box)
    ax.text(x, y, text, ha='center', va='center',
            fontsize=fontsize, fontweight='bold' if bold else 'normal',
            color=C_TXT, wrap=True,
            multialignment='center')

def draw_box_dashed(ax, x, y, w, h, text, fontsize=9):
    from matplotlib.patches import Rectangle
    rect = Rectangle((x - w/2, y - h/2), w, h,
                     facecolor='#F5F5FF', edgecolor=C_DASH,
                     linewidth=0.8, linestyle='--')
    ax.add_patch(rect)
    ax.text(x, y + h/2 - 0.04, text, ha='center', va='top',
            fontsize=7.5, color=C_DASH, style='italic')

def arrow(ax, x1, y1, x2, y2, dashed=False):
    style = 'dashed' if dashed else 'solid'
    ax.annotate('', xy=(x2, y2), xytext=(x1, y1),
                arrowprops=dict(
                    arrowstyle='->', color=C_ARROW,
                    lw=0.9, linestyle=style,
                    connectionstyle='arc3,rad=0'))

def arrow_label(ax, x1, y1, x2, y2, label, dashed=False):
    arrow(ax, x1, y1, x2, y2, dashed)
    mx, my = (x1+x2)/2, (y1+y2)/2
    ax.text(mx+0.02, my, label, ha='left', va='center', fontsize=7.5, color=C_DASH)

def save(fig, fname):
    fig.savefig(f'{OUT}/{fname}')
    plt.close()
    print(f"✅ {fname}")
```

```python
# ══════════════════════════════════════════════════════════════════════════════
# GAMBAR 3 — Pipeline Preprocessing (12 tahap)
# ══════════════════════════════════════════════════════════════════════════════
def gambar3_preprocessing():
    fig, ax = plt.subplots(figsize=(13, 5))
    ax.set_xlim(0, 13); ax.set_ylim(0, 5); ax.axis('off')
    ax.set_title('Gambar 3 Pipeline Preprocessing',
                 fontsize=11, fontweight='bold', pad=8)

    W, H = 2.7, 0.65

    # Row 1 (kiri→kanan): 1→2→3→4
    row1 = [(1.5,3.8,'1. Duplication\nRemover'),
            (4.4,3.8,'2. URL/At/Hash\nRemover'),
            (7.3,3.8,'3. Emoji→Sentiment\nConversion'),
            (10.2,3.8,'4. Punctuation\nRemover')]
    for x,y,lbl in row1:
        draw_box(ax, x, y, W, H, lbl, fontsize=8.5)
    for i in range(len(row1)-1):
        arrow(ax, row1[i][0]+W/2, row1[i][1], row1[i+1][0]-W/2, row1[i+1][1])
    # Turn kanan→bawah
    arrow(ax, row1[-1][0], row1[-1][1]-H/2, row1[-1][0], row1[-1][1]-H/2-0.45)

    # Row 2 (kanan→kiri): 8→7→6→5
    row2 = [(10.2,2.5,'8. Slang Correction\n/ Standardization'),
            (7.3,2.5,'7. Double Letter\nRemoval'),
            (4.4,2.5,'6. Tokenization'),
            (1.5,2.5,'5. Case Folding')]
    for x,y,lbl in row2:
        draw_box(ax, x, y, W, H, lbl, fontsize=8.5)
    for i in range(len(row2)-1):
        arrow(ax, row2[i][0]-W/2, row2[i][1], row2[i+1][0]+W/2, row2[i+1][1])
    # Turn kiri→bawah
    arrow(ax, row2[-1][0], row2[-1][1]-H/2, row2[-1][0], row2[-1][1]-H/2-0.45)

    # Row 3 (kiri→kanan): 9→10→11→12
    row3 = [(1.5,1.2,'9. Stopword\nRemoval'),
            (4.4,1.2,'10. Stemming\n(PySastrawi+cache)'),
            (7.3,1.2,'11. KBBI Filter'),
            (10.2,1.2,'12. Final Filter')]
    for x,y,lbl in row3:
        draw_box(ax, x, y, W, H, lbl, fontsize=8.5)
    for i in range(len(row3)-1):
        arrow(ax, row3[i][0]+W/2, row3[i][1], row3[i+1][0]-W/2, row3[i+1][1])

    # # Input label
    # ax.text(0.1, 3.8, 'Input\n80.163', ha='center', va='center',
    #         fontsize=8, color='#555555', style='italic')
    # arrow(ax, 0.3, 3.8, 1.5-W/2, 3.8)

    # # Output
    # arrow(ax, row3[-1][0]+W/2, row3[-1][1], 12.2, row3[-1][1])
    # draw_box(ax, 12.7, 1.2, 1.0, 0.65, '70.135\nbaris', fontsize=8, fc='#EEF4FF')

    fig.tight_layout()
    save(fig, 'Gambar03_pipeline_preprocessing.png')
```
